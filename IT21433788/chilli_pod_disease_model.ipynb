```python
from ultralytics import YOLO
from google.colab import files

# Upload test image
print("Upload test image")
uploaded = files.upload()
test_image = list(uploaded.keys())[0]

# Load model with correct path
model = YOLO('/content/runs/classify/chilli_disease_model/weights/best.pt')

# Predict
results = model.predict(test_image)
top_class = results[0].names[results[0].probs.top1]
confidence = results[0].probs.top1conf.item()

print(f"Predicted Disease: {top_class}")
print(f"Confidence: {confidence:.2f}")
```

```python
# --- Serverity & Treatment ---

treatment_db = {
    "Anthracnose": {
        "medicine": "Mancozeb 80% WP",
        "dosage": "2.5g per 1L water",
        "method": "Remove infected pods & Spray every 7 days"
    },
    "Phytophthora": {
        "medicine": "Metalaxyl + Mancozeb (Ridomil Gold)",
        "dosage": "2g per 1L water",
        "method": "Avoid waterlogging & Spray at base"
    },
    "Bacterial_Spot": {
        "medicine": "Copper Hydroxide",
        "dosage": "2.5g per 1L water",
        "method": "Remove infected leaves first"
    },
    "Healthy": {
        "medicine": "Neem Oil (Prevention)",
        "dosage": "2ml per 1L water",
        "method": "Spray every 14 days"
    }
}

import cv2
import numpy as np

def calculate_severity(image_path):
    img = cv2.imread(image_path)
    hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
    lower = np.array([10, 40, 30])
    upper = np.array([30, 255, 200])
    mask = cv2.inRange(hsv, lower, upper)
    severity_percent = (np.count_nonzero(mask) / mask.size) * 100 * 2.5
    severity_percent = min(max(severity_percent, 1.0), 95.0)

    if severity_percent < 10: level = "Low"
    elif severity_percent < 40: level = "Medium"
    else: level = "High"
    return level, severity_percent

severity_level, severity_per = calculate_severity(test_image)
info = treatment_db.get(top_class, treatment_db["Healthy"])

print("="*45)
print(f"🌶️ DISEASE: {top_class}")
print(f"📊 SEVERITY: {severity_level} ({severity_per:.1f}%)")
print(f"💊 MEDICINE: {info['medicine']}")
print(f"🧪 DOSAGE: {info['dosage']}")
print(f"📝 METHOD: {info['method']}")
print("="*45)
```
